```python
import random
from pathlib import Path
from dotenv import load_dotenv
```

```python
from langchain.tools import tool
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain.agents import create_agent
```

```python
from datetime import datetime
```

```python
# Carregando as variáveis de ambiente
ENV_PATH = Path('../.env')
load_dotenv(dotenv_path=ENV_PATH)
```

```python
@tool
def throw_dice_tool() -> str:
    """You roll a die and get a person's name."""
    DICE = {
        'ONE': 'Luiz',
        'TWO': 'Two',
        'THREE': 'Nuna',
        'FOUR': 'Henrique',
        'FIVE': 'Egito',
        'SIX': 'Luiz Henrique'
    }
    number = random.choice(list(DICE.keys()))
    return f'O numero do dado foi: {number}, e o nome escolhido foi: {DICE[number]}'
```

```python
@tool
def current_datetime() -> dict:
    """Returns the current date and time."""
    now = datetime.now()
    return {
        'year': now.year,
        'month': now.month,
        'day': now.day,
        'hour': now.hour,
        'minute': now.minute,
        'second': now.second
    }
```

```python
llm = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0
)
```

```python
TOOLS = [throw_dice_tool, current_datetime]
```

```python
agent = create_agent(
    model=llm,
    tools=TOOLS,
    system_prompt="Você é um assistente útil."
)
```

```python
prompt = '''
Lance um dado e me o seu número e o nome que o dado escolheu por favor.
Além disso me diga a data horas minutos e segundos que você jogou o dado,
por favor.
'''
response = agent.invoke({"input": prompt})
```
