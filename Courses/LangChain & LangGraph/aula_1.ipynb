```python
from rich import print as pprint
from langchain.chat_models import init_chat_model
from langchain_core.messages import SystemMessage, HumanMessage
```

```python
llm = init_chat_model('ollama:gemma:2b')
llm
```

```python
response = llm.invoke('Olá, tudo bem?')
pprint(response)
```

```python
system_message = SystemMessage('Você deve ser curto e suscinto nas suas respostas.')
# esse é seu prompt
human_message = HumanMessage('Olá tudo bem?')

messages = [
    system_message,
    human_message
]

response = llm.invoke(messages)
pprint(response)
```

```python
# iniciando uma conversa
system_message = SystemMessage('Você só pode responder em inglês nunca em outro idioma.')
# esse é seu prompt
human_message = HumanMessage('Olá tudo bem?')

messages = [
    system_message,
    human_message
]

response = llm.invoke(messages)
print(f'{"AI":-^80}')
print(response.content)

messages.append(response)

while True:
    pprint(f'{"Human":-^80}')
    user_input = input('[Message] > ')
    human_message = HumanMessage(user_input)

    if user_input.lower() in ['exit', 'quit', 'q']:
        break

    messages.append(human_message)

    response = llm.invoke(messages)

    print(f'{"AI":-^80}')
    print(response.content)
    messages.append(response)
```
