```python
import random
from smolagents import tool, CodeAgent, OpenAIModel
```

```python
from pathlib import Path
from dotenv import load_dotenv
```

```python
from datetime import datetime
```

```python
# Carregando as variáveis de ambiente
ENV_PATH = Path('../.env')
load_dotenv(dotenv_path=ENV_PATH)
```

```python
@tool
def throw_dice_tool() -> int:
    'You roll a die and get a persons name.'
    DICE = {
        'ONE': 'Luiz',
        'TWO': 'Two',
        'THREE': 'Nuna',
        'FOUR': 'Henrique',
        'FIVE': 'Egito',
        'SIX': 'Luiz Henrique'
    }
    number = random.choice(list(DICE.keys()))
    return f'O numero do dado foi: {number}, e o nome escolhido foi: {DICE[number]}'
```

```python
model = OpenAIModel(
    model_id="gpt-4o",
)

agent = CodeAgent(tools=[throw_dice_tool], model=model, verbosity_level=0)

result = agent.run("Jogue o dado e me diga qual nome da pessoa e qual numero ele caiu.")
print(result)
```

# Adicionando mais de uma Tool

```python
# criando uma nova tool
@tool
def current_datetime() -> dict[str, int]:
    "I'll give you back the exact year, month, day, minutes, and seconds."
    now = datetime.now()
    return {
        'year': now.year,
        'month': now.month,
        'day': now.day,
        'hour': now.hour,
        'minute': now.minute,
        'second': now.second
    }
```

```python
TOOLS = [current_datetime, throw_dice_tool]
```

```python
model = OpenAIModel(
    model_id="gpt-4o",
)

agent = CodeAgent(tools=TOOLS, model=model, verbosity_level=0)

prompt = '''
Lance um dado e me o seu número e o nome que o dado escolheu por favor.
Além disso me diga a data horas minutos e segundos que você jogou o dado,
por favor.
'''

result = agent.run(prompt)
print(result)
```
